Start JupyterLab and Trail as in [setup](setup.md), then open this notebook in `trail-tutorial`. JupyterLab is the browser workspace; JupyterGIS is its map editor. The editor reports actions through Jupyter Server, where Trail receives them.

## Open, switch and close maps

Open `maps/earthquakes.jGIS` from the file browser. Switch between it and the new map, then close the new map's tab.

| Action | Trail records |
|---|---|
| Open an existing map | `opened`, then `activated`. |
| Switch map tabs | `deactivated` for the old tab and `activated` for the new one. |
| Close a map tab | `deactivated`, then `closed`. |

These are document events. Each names the map in `src_path`. Focus events also identify the browser page (`frontend_id`) and tab (`widget_id`).

## Add a layer

A source provides data; a layer draws it on the map.

In the map toolbar, open **Add Layer** and choose **GeoJSON**. Set the path to `../data/crossings.geojson` and confirm. File paths here are relative to the map, which is in `maps/`.

Trail records a source added and a layer added. Both link to the tracked data file:

| Field | Meaning |
|---|---|
| `resource_path` | The data file's path within the Trail project. |
| `resource_id` | Its tracked asset id. |
| `local` | Whether the edit came from the browser tab reporting it. |

An untracked file is marked `resource: untracked`. Sources containing a URL or inline data have no tracked file link.

## What the saved map shows

Save the map.

Live map events describe editor actions. File records describe what reached disk: Trail compares the saved map's layers, sources, order and settings with its previous saved state.

Consecutive server saves are grouped into a `ContentsServiceV1Event` after 30 seconds without another save. Structural changes can appear in file records or a separate `MetadataEvent`. See [file activity](files.ipynb).


## Query map history

Run these in Trail:

```text
events cls=LayerEvent src_path=maps/earthquakes.jGIS
events resource_path=data/crossings.geojson
events cls=LayerEvent has_filter_change=true
events cls=ModeEvent mode=identifying
events cls=ViewportEvent src_path=maps/earthquakes.jGIS -5:
```
